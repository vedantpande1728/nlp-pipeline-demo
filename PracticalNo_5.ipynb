```python
import numpy as np

print("--- Simple Attention Mechanism Explanation ---")

# 1. Dummy Encoder Outputs (e.g., representations of input words)
# Imagine these are vector representations of words like 'I', 'am', 'a', 'student'
encoder_outputs = np.array([
    [0.1, 0.2],  # 'I'
    [0.5, 0.3],  # 'am'
    [0.2, 0.8],  # 'a'
    [0.9, 0.1]   # 'student'
])
print(f"\nEncoder Outputs (4 words, 2 features each):\n{encoder_outputs}")

# 2. Dummy Decoder Hidden State (what the decoder is currently trying to generate)
# Let's say the decoder is trying to generate a word related to 'learning'
decoder_hidden_state = np.array([0.7, 0.4])
print(f"\nDecoder Hidden State (what we are looking for): {decoder_hidden_state}")

# 3. Calculate 'Alignment Scores' or 'Energy' (how well each encoder output matches the decoder state)
# A simple dot product can show similarity
alignment_scores = np.dot(encoder_outputs, decoder_hidden_state)
print(f"\nAlignment Scores (similarity of each encoder output to decoder state): {alignment_scores}")

# 4. Apply Softmax to get 'Attention Weights' (probabilities summing to 1)
def softmax(x):
    e_x = np.exp(x - np.max(x)) # Subtracting max for numerical stability
    return e_x / e_x.sum(axis=0)

attention_weights = softmax(alignment_scores)
print(f"\nAttention Weights (how much to 'pay attention' to each word): {attention_weights}")
print(f"(Notice that the word 'student' (index 3) got a higher weight, meaning it's more relevant to 'learning')")

# 5. Compute the 'Context Vector' (a weighted sum of encoder outputs)
# This is what the decoder actually uses, focusing on the most relevant parts
context_vector = np.sum(encoder_outputs * attention_weights[:, np.newaxis], axis=0)
print(f"\nContext Vector (weighted sum of encoder outputs, 'paying attention'): {context_vector}")

print("\n--- How it works ---")
```
